# 🐍 Python Functions — 4 tips to write better Python Functions

These are 4 tips to writing better Python Functions

## 1. Specify Parameter Values

**Description:**  
Python functions can accept parameters without telling you what kind of values they expect. Adding type hints to parameters makes the function contract clear and helps both readers and tools understand how to use the function correctly.

`get_user_count(users: dict[int, str]) -> int`

This function expects a dictionary where:
- the keys are integers
- the values are strings

The parameter annotation `users: dict[int, str]` tells Python and anyone reading the code that this function must receive a dictionary of user IDs to names. This makes the function easier to use correctly and reduces mistakes.

Without this annotation, the caller may not know that `users` should be a dictionary rather than a list or a string.

`display_users(users: dict[int, str]) -> None`

This function also expects a dictionary of user IDs and names. It loops through the items and prints them to the console.

The parameter type makes it clear that the function is designed to work with structured data, not arbitrary values.

---

## 2. Specify Return Types

**Description:**  
Python is dynamically typed, so the function signature alone does not tell you what it returns. Adding a return type makes the intent explicit.

`get_users() -> dict[int, str]`

This function returns a dictionary where:
- the keys are integers
- the values are strings

The annotation `-> dict[int, str]` tells anyone reading the code, and any static analysis tool, exactly what kind of value to expect. Without this type, Python would not make it clear that the function is supposed to return a dictionary of user IDs to names.

`get_user_count() -> int`

This function calls `get_users()` and then returns `len(users)`. Since the length of a dictionary is always a number, the return type is clearly `int`.

This makes the function contract explicit: “I return the number of users.”

`display_users() -> None`

This function prints each user from the dictionary returned by `get_users()`. It loops through the dictionary items and displays them in a readable format:

`User ID: {user_id}, User Name: {user_name}`

The return type `-> None` means that the function does not return a value. Instead, it performs a side effect by printing the users to the console.

This is a useful pattern when a function is meant to display information, rather than calculate or return data.

---
## 3. Add a Docstring

**Description:**  
A docstring is a short description placed inside a function that explains what the function does, what arguments it expects, and what it returns. It is one of the easiest ways to document your code clearly.

```python
def get_users() -> dict[int, str]:
    """
    :return: A dictionary of user IDs and names.
    """
    return {1: "Bob", 2: "Jeff", 3: "Tom"}
```

Docstrings are useful because they make the function easier to understand for other developers, and many IDEs will show this information when you hover over the function name or call it. In VS Code, for example, hovering over `get_users()` can display the docstring, helping you understand the function without reading the implementation.

This makes the code self-documenting and improves readability, maintenance, and collaboration.

````python
def get_users() -> dict[int, str]:
    """
    :return: A dictionary of user IDs and names.
    """
    return {1: "Bob", 2: "Jeff", 3: "Tom"}


def get_user_count(users: dict[int, str]) -> int:
    """
    :param users: A dictionary of user IDs and names.
    :return: The number of users.
    """
    return len(users)


def display_users(users: dict[int, str]) -> None:
    """
    :param users: A dictionary of user IDs and names.
    """
    for user_id, user_name in users.items():
        print(f"User ID: {user_id}, User Name: {user_name}")


users = get_users()
display_users(users)
print(get_user_count(users))
````
