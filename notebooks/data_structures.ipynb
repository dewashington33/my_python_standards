# 🐍 Python Data Structures — Markdown Outline

## 1. **Lists**
**Description:**  
Ordered, mutable collections that can hold mixed data types. Ideal for sequences where items may change.

```python
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4]
fruits.append("orange")
numbers.remove(2)
print(fruits)
print(numbers)
```

## 2. **Tuples**
**Description:**  
Ordered, **immutable** collections. Great for fixed data that shouldn’t change.

```python
coordinates = (33.5, -84.4)
person = ("David", 55, "Perry, GA")
print(coordinates)
print(person)
```

## 3. **Sets**
**Description:**  
Unordered collections of **unique** items. Useful for membership tests and removing duplicates.

```python
unique_numbers = {1, 2, 3, 3, 2}   # becomes {1, 2, 3}
print(unique_numbers)
unique_numbers.add(4)
print(unique_numbers)
```

## 4. **Dictionaries**
**Description:**  
Key–value mappings. Mutable, unordered, and extremely flexible for structured data.


```python
person = {"name": "David", "city": "Perry", "role": "DBA"}
person["role"] = "Museum Governance Architect"
print(person)
```

## 5. **Strings**
**Description:**  
Immutable sequences of characters. Technically a data structure because they behave like sequences.


```python
message = "Hello World"
print(message[0])   # 'H'
print(message[6])   # 'W'
```

## 6. **Arrays (`array` module)**
**Description:**  
More memory‑efficient than lists, but only hold one data type. Often used for numeric storage.


```python
# Python does not have a built-in array data structure like some other languages, 
# but it does have a module called `array` that provides an array type that is more 
# efficient than lists for certain use cases. Here's an example of how to use the `array` module:
from array import array
# This is a typed array of integers
nums = array('i', [1, 2, 3])
nums.append(4)
print(nums)
```

## 7. **Queues (`collections.deque`)**
**Description:**  
Double‑ended queue with fast appends/pops from both ends. Perfect for FIFO or LIFO operations.

```python
from collections import deque
q = deque([1, 2, 3])
q.append(4)
print(q.popleft())   # FIFO
print(q)             # deque([2, 3, 4])
```

## 8. **Stacks (via list or deque)**
**Description:**  
LIFO structure. Python doesn’t have a dedicated stack type; lists or deques are used.

```python
stack = []
stack.append("task1")
stack.append("task2")
print(stack)   # ['task1', 'task2']
stack.pop()   # removes "task2"
print(stack)   # ['task1']
```

## 9. **Named Tuples**
**Description:**  
Tuples with named fields. Immutable and more readable than regular tuples.

** Named tuples are not commonly used.  A more modern Python approach would be to use a dataclass.  See example of a dataclass next.

```python
from collections import namedtuple
Point = namedtuple("Point", "x y")
p1 = Point(10, 20)
p2 = Point(30, 40)
p3 = Point(50, 60)
print("Points stored in namedtuples:")
print(p1.x, p1.y)   # 10 20
print(p2.x, p2.y)  # 30 40
print(p3.x, p3.y)  # 50 60

# To store them together use a list or a dictionary
print("Points stored in a list:")
points = [p1, p2, p3]
for point in points:
    print(point.x, point.y)
```

## 10. **Data Classes**
**Description:**  
A dataclass is a class designed to store related data. Python automatically provides methods such as an initializer and a readable string representation.

```python
from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int

point1 = Point(10, 20)
point2 = Point(30, 40)
point3 = Point(50, 60)
print("Points stored in dataclasses:")
print(point1.x, point1.y)  # 10 20
print(point2.x, point2.y)  # 30 40
print(point3.x, point3.y)  # 50 60

# To store them together use a list or a dictionary 
points = [point1, point2, point3]
print("Points stored in a list:")
for point in points:
    print(point.x, point.y)
```
