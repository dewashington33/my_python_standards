# 🐍 Python `collections.abc` — Commonly Used ABCs

These are the abstract base classes most Python developers use regularly when checking behavior, writing reusable code, or understanding built-in collection types.


## 1. **Iterable**
**Description:**  
An object that can be looped over with `for`. Lists, tuples, strings, sets, dictionaries, and files are all iterable.


```python
from collections.abc import Iterable

names = ["Alice", "Bob", "Charlie"]
message = "hello"
numbers = {1, 2, 3}

print(isinstance(names, Iterable))
print(isinstance(message, Iterable))
print(isinstance(numbers, Iterable))

for name in names:
    print(name)
```

## 2. **Iterator**
**Description:**  
An iterator gives one value at a time using `next()`. It is created from an iterable with `iter()`.


```python
from collections.abc import Iterator

scores = [88, 92, 97]
score_iter = iter(scores)

print(type(score_iter))
print(isinstance(score_iter, Iterator))
print(next(score_iter))
print(next(score_iter))
print(next(score_iter))
```

## 3. **Generator**
**Description:**  
A generator is a special kind of iterator created with `yield`. It is useful for streaming values without storing them all in memory.


```python
from collections.abc import Generator

def invoice_numbers(limit):
    for i in range(1, limit + 1):
        yield f"INV-{i:04d}"

invoices = invoice_numbers(5)
print(isinstance(invoices, Generator))
print(next(invoices))
print(next(invoices))
print(list(invoices))
```

## 4. **Sequence**
**Description:**  
A sequence is an ordered collection with indexing and slicing. Common examples are lists, tuples, and strings.


```python
from collections.abc import Sequence

order = ["milk", "eggs", "bread"]
name = "David"

print(isinstance(order, Sequence))
print(isinstance(name, Sequence))
print(order[0])
print(order[1:])
print(name[0:3])
```

## 5. **MutableSequence**
**Description:**  
A mutable sequence allows items to be added, removed, or replaced. Lists are the main example.


```python
from collections.abc import MutableSequence

shopping_list = ["milk", "eggs"]
print(isinstance(shopping_list, MutableSequence))

shopping_list.append("bread")
shopping_list.insert(1, "fruit")
shopping_list.remove("milk")
print(shopping_list)
```

## 6. **Set**
**Description:**  
A set is an unordered collection of unique items. Great for membership tests and removing duplicates.


```python
from collections.abc import Set

customers = {"Alice", "Bob", "Charlie"}
vip_customers = {"Alice", "Charlie"}

print(isinstance(customers, Set))
print(customers | vip_customers)
print(customers & vip_customers)
print(customers - vip_customers)
print("Alice" in customers)
```

## 7. **MutableSet**
**Description:**  
A mutable set lets you add or remove values after creation. This is useful when tracking a changing collection of unique items.


```python
from collections.abc import MutableSet

active_users = {"david", "laura"}
print(isinstance(active_users, MutableSet))

active_users.add("mike")
active_users.discard("laura")
print(active_users)
```

## 8. **Mapping**
**Description:**  
A mapping stores key-value pairs, like a dictionary. It lets you look up data by name instead of index.


```python
from collections.abc import Mapping

employee = {
    "name": "David",
    "city": "Perry",
    "role": "DBA"
}

print(isinstance(employee, Mapping))
print(employee["name"])
print(list(employee.keys()))
print(list(employee.values()))
print(list(employee.items()))
```

## 9. **MutableMapping**
**Description:**  
A mutable mapping supports adding, updating, and deleting key-value pairs. This is the standard behavior of `dict`.


```python
from collections.abc import MutableMapping

person = {"name": "David"}
print(isinstance(person, MutableMapping))

person["city"] = "Perry"
person["role"] = "DBA"
person["name"] = "David W. Smith"
person.pop("city")
print(person)
```

## 10. **Real-World Summary**
**Description:**  
These ABCs show how Python organizes common collection behavior:
- `Iterable`: can be looped over
- `Iterator`: produces values one at a time
- `Generator`: lazy iterator built with `yield`
- `Sequence`: ordered and indexable
- `MutableSequence`: ordered and changeable
- `Set`: unique unordered values
- `MutableSet`: changeable unique values
- `Mapping`: key-value data
- `MutableMapping`: key-value data that can be changed

These are the big ideas most developers use every day.


```python
from collections.abc import Iterable, Sequence, Set, Mapping

samples = {
    "shopping_list": ["milk", "eggs", "bread"],
    "employee": {"name": "David", "role": "DBA"},
    "active_users": {"alice", "bob", "charlie"},
    "word": "hello",
}

for name, value in samples.items():
    print(name)
    print("  Iterable:", isinstance(value, Iterable))
    print("  Sequence:", isinstance(value, Sequence))
    print("  Set:", isinstance(value, Set))
    print("  Mapping:", isinstance(value, Mapping))
```
