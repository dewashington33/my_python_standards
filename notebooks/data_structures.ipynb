# 🐍 Python Data Structures — Markdown Outline

## 1. **Lists**
**Description:**  
Ordered, mutable collections that can hold mixed data types. Ideal for sequences where items may change.

```python
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4]
fruits.append("orange")
numbers.remove(2)
print(fruits)
print(numbers)
```

## 2. **Tuples**
**Description:**  
Ordered, **immutable** collections. Great for fixed data that shouldn’t change.

```python
coordinates = (33.5, -84.4)
person = ("David", 55, "Perry, GA")
print(coordinates)
print(person)
```

## 3. **Sets**
**Description:**  
Unordered collections of **unique** items. Useful for membership tests and removing duplicates.

```python
unique_numbers = {1, 2, 3, 3, 2}   # becomes {1, 2, 3}
print(unique_numbers)
unique_numbers.add(4)
print(unique_numbers)
```

## 4. **Dictionaries**
**Description:**  
Key–value mappings. Mutable, unordered, and extremely flexible for structured data.


```python
person = {"name": "David", "city": "Perry", "role": "DBA"}
person["role"] = "Museum Governance Architect"
print(person)
```
